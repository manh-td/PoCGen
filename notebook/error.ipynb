```python
from pathlib import Path
import json

result_path = Path("../output")

json_files = list(result_path.glob('**/RunnerResult_DefaultRefiner.json'))
data = []
for file in json_files:
    try:
        datapoint = json.load(open(file))
        data.append(datapoint)
    except Exception as e:
        print(file)
len(data)
```

```python
error_map = {
    "Timeout": [],
    "Prompt tokens exceeded": [],
    "Completion tokens exceeded": [],
    "Advisory not found or removed": [],
    "Could not create codeql database": [],
    "Unexpected token": [],
    "Connection error.": [],
    "Could not initialize npm package": []
}

# Collect timeout kills from pipeline result files
pipeline_files = list(result_path.glob('DefaultRefiner_pipeline_*.json'))
seen_timeout_ids = set()
for pf in pipeline_files:
    pipeline_data = json.load(open(pf))
    for entry in pipeline_data.values():
        if not isinstance(entry, dict):
            continue
        if "Exited with code" in str(entry.get("error", "")):
            id = entry.get("advisoryId", str(pf))
            if id not in seen_timeout_ids:
                error_map["Timeout"].append(id)
                seen_timeout_ids.add(id)

# Collect runner-level errors from RunnerResult files
datapoint: dict
for datapoint in data:
    if "error" in datapoint:
        id = datapoint['baseDir'].replace('/app/output/', '').replace('_', ':')
        message = datapoint['error']['message']
        if "Prompt tokens exceeded" in message:
            error_map["Prompt tokens exceeded"].append(id)
        elif "Completion tokens exceeded" in message:
            error_map["Completion tokens exceeded"].append(id)
        elif "Advisory not found or removed" in message:
            error_map["Advisory not found or removed"].append(id)
        elif "Could not create codeql database" in message:
            error_map["Could not create codeql database"].append(id)
        elif "Unexpected token" in message:
            error_map["Unexpected token"].append(id)
        elif "Connection error." in message:
            error_map["Connection error."].append(id)
        elif "Could not initialize npm package" in message:
            error_map["Could not initialize npm package"].append(id)
        else:
            print(id)
            print(message)
            break

# Print summary with counts
for k, v in error_map.items():
    print(f"{k}: {len(v)}")
error_map
```
